# Relationship Analysis

Agricultural performance is influenced by multiple environmental conditions, farming inputs, and economic factors.

After comparing agricultural performance across seasons, crops, and geographical regions, this section investigates relationships between important variables and agricultural outcomes.

The analysis focuses on relationships between:

- Environmental conditions and crop yield
- Resource usage and agricultural production
- Farm characteristics and production
- Economic variables and profit

```python
plt.figure(figsize=(9, 6))

sns.scatterplot(
    data=df,
    x='Rainfall_mm',
    y='Yield_Tonnes_Ha',
    hue='Season',
    hue_order=['Kharif', 'Rabi', 'Zaid'],
    alpha=0.6
)

plt.title("Relationship Between Rainfall and Crop Yield")
plt.xlabel("Rainfall (mm)")
plt.ylabel("Yield (Tonnes per Hectare)")

plt.legend(title="Season")

plt.tight_layout()
plt.show()
```

```python
rainfall_yield_corr = df['Rainfall_mm'].corr(
    df['Yield_Tonnes_Ha']
)
print(
    "Correlation between Rainfall and Yield:",
    round(rainfall_yield_corr, 3)
)
```

#  Key Findings and Insights

This section summarizes the major patterns and observations identified during the seasonal agricultural performance analysis.

The findings are based on the analysis of agricultural production, environmental conditions, resource usage, economic performance, crop-wise patterns, regional variations, and relationships between important variables.

```python
seasonal_ranking = overall_performance.copy()

# Rank based on performance indicators
seasonal_ranking['Yield_Rank'] = (
    seasonal_ranking['Yield_Tonnes_Ha']
    .rank(ascending=False, method='dense')
)

seasonal_ranking['Production_Rank'] = (
    seasonal_ranking['Production_Tonnes']
    .rank(ascending=False, method='dense')
)

seasonal_ranking['Water_Efficiency_Rank'] = (
    seasonal_ranking['Water_Efficiency_t_per_1000m3']
    .rank(ascending=False, method='dense')
)

seasonal_ranking['Profit_Rank'] = (
    seasonal_ranking['Profit_INR']
    .rank(ascending=False, method='dense')
)

seasonal_ranking
```

```python
performance_comparison = overall_performance.copy()

performance_columns = [
    'Yield_Tonnes_Ha',
    'Production_Tonnes',
    'Water_Efficiency_t_per_1000m3',
    'Profit_INR'
]

normalized_performance = (
    performance_comparison[performance_columns]
    - performance_comparison[performance_columns].min()
) / (
    performance_comparison[performance_columns].max()
    - performance_comparison[performance_columns].min()
)

normalized_performance
```

```python
normalized_performance.plot(
    kind='bar',
    figsize=(12, 6)
)

plt.title('Normalized Comparison of Agricultural Performance Across Seasons')
plt.xlabel('Season')
plt.ylabel('Normalized Performance Score')

plt.xticks(rotation=0)

plt.legend(
    [
        'Yield',
        'Production',
        'Water Efficiency',
        'Profit'
    ],
    title='Performance Indicator'
)

plt.tight_layout()
plt.show()
```
